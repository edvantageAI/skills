# VideoRAG Installation Tutorial

This notebook guides you through installing and configuring VideoRAG for the YouTube Lecture Processor skill.

**Estimated time:** 30-45 minutes (depending on download speeds)

**Prerequisites:**
- Python 3.9 or higher
- Git and Git LFS installed
- 15GB free disk space
- GPU with CUDA support (recommended) or CPU

## Step 1: System Requirements Check

First, let's verify your system meets the requirements.

````python
import sys
import subprocess

# Check Python version
print(f"Python version: {sys.version}")
assert sys.version_info >= (3, 9), "Python 3.9 or higher required"

# Check Git
try:
    git_version = subprocess.check_output(['git', '--version']).decode().strip()
    print(f"✓ {git_version}")
except:
    print("✗ Git not found. Install from: https://git-scm.com/")

# Check Git LFS
try:
    lfs_version = subprocess.check_output(['git', 'lfs', 'version']).decode().strip()
    print(f"✓ {lfs_version}")
except:
    print("✗ Git LFS not found. Install from: https://git-lfs.com/")
````

## Step 2: Clone VideoRAG Repository

Clone the VideoRAG repository from GitHub.

````python
import os

# Set installation directory
INSTALL_DIR = os.path.expanduser("~/videorag")

# Clone repository
if not os.path.exists(INSTALL_DIR):
    !git clone https://github.com/HKUDS/VideoRAG.git {INSTALL_DIR}
    print(f"✓ Cloned VideoRAG to {INSTALL_DIR}")
else:
    print(f"✓ VideoRAG already exists at {INSTALL_DIR}")

# Change to VideoRAG directory
os.chdir(INSTALL_DIR)
print(f"Working directory: {os.getcwd()}")
````

## Step 3: Set Up Python Environment with pyenv

Create a virtual environment using pyenv and venv.

````python
# Check if pyenv is installed
try:
    pyenv_version = subprocess.check_output(['pyenv', '--version']).decode().strip()
    print(f"✓ {pyenv_version}")
except:
    print("✗ pyenv not found. Install from: https://github.com/pyenv/pyenv")
    print("Falling back to system Python...")

# Create virtual environment
venv_path = os.path.join(INSTALL_DIR, "venv")
if not os.path.exists(venv_path):
    !python -m venv {venv_path}
    print(f"✓ Created virtual environment at {venv_path}")
else:
    print(f"✓ Virtual environment already exists at {venv_path}")

print("\nTo activate the environment, run:")
print(f"  source {venv_path}/bin/activate  # Linux/Mac")
print(f"  {venv_path}\\Scripts\\activate  # Windows")
````

## Step 4: Install Dependencies with uv

Install uv (fast Python package installer) and use it to install dependencies.

**Note:** Make sure you've activated the virtual environment before running this cell.

````python
# Install uv
!pip install uv

# Install PyTorch (GPU version)
print("Installing PyTorch with CUDA support...")
!uv pip install torch torchvision torchaudio --index-url https://download.pytorch.org/whl/cu118

# Install VideoRAG dependencies
print("\nInstalling VideoRAG dependencies...")
!uv pip install transformers accelerate bitsandbytes sentencepiece protobuf

# Install LLM provider SDKs
print("\nInstalling LLM provider SDKs...")
!uv pip install openai anthropic google-generativeai

# Install video processing libraries
print("\nInstalling video processing libraries...")
!uv pip install opencv-python ffmpeg-python yt-dlp

# Install additional utilities
!uv pip install numpy pandas tqdm pillow

print("\n✓ All dependencies installed!")
````

Download the required models (~10GB). This may take 15-30 minutes depending on your internet speed.

````python
# Initialize Git LFS
!git lfs install

# Pull model files
print("Downloading models (this may take 15-30 minutes)...")
!git lfs pull

print("\n✓ Models downloaded successfully!")
````

## Step 7: Verify Installation

Test that everything is working correctly.

````python
# Test imports
print("Testing imports...")

try:
    import torch
    print(f"✓ PyTorch {torch.__version__}")
    print(f"  CUDA available: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"  CUDA version: {torch.version.cuda}")
        print(f"  GPU: {torch.cuda.get_device_name(0)}")
except ImportError as e:
    print(f"✗ PyTorch import failed: {e}")

try:
    import transformers
    print(f"✓ Transformers {transformers.__version__}")
except ImportError as e:
    print(f"✗ Transformers import failed: {e}")

try:
    import cv2
    print(f"✓ OpenCV {cv2.__version__}")
except ImportError as e:
    print(f"✗ OpenCV import failed: {e}")

try:
    import openai
    print(f"✓ OpenAI SDK {openai.__version__}")
except ImportError as e:
    print(f"✗ OpenAI SDK import failed: {e}")

print("\n✓ Installation verification complete!")
````

## Troubleshooting

### CUDA Out of Memory
- Reduce batch size in VideoRAG config
- Use smaller models
- Close other GPU applications

### Git LFS Bandwidth Limit
- Wait 1 hour for quota reset
- Use a different network
- Download models manually from Hugging Face

### Import Errors
- Ensure virtual environment is activated
- Reinstall with: `uv pip install --force-reinstall <package>`
- Check Python version compatibility

### Slow Performance
- Verify GPU is being used: `torch.cuda.is_available()`
- Update CUDA drivers
- Use smaller models or CPU-only mode
